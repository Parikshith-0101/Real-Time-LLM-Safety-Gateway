```python

in_path = "llm_prompts_dataset.csv"
out_path = "clean_llm_dataset.csv"

# Read only the required columns
try:
    df = pd.read_csv(in_path, usecols=["prompt", "attack_type"])
except Exception:
    # fallback: read full CSV then select columns (handles files without header or different order)
    df = pd.read_csv(in_path)
    df = df[[c for c in ["prompt", "attack_type"] if c in df.columns]]

total_before = len(df)

# Normalize attack_type: replace NaN with S, strip whitespace, title-case
# Normalize attack_type: replace NaN with "S", strip whitespace, title-case
if "attack_type" in df.columns:
    df["attack_type"] = df["attack_type"].fillna("S")  # <-- replace NaN with S
    df["attack_type"] = df["attack_type"].astype(str).str.strip()
    df["attack_type"] = df["attack_type"].apply(lambda x: x.title() if x != "S" else "S")
else:
    df["attack_type"] = "S"


# Clean prompt: drop rows with missing/empty prompts after stripping
df["prompt"] = df["prompt"].astype(str).str.strip()
df = df[df["prompt"] != ""]

# Drop exact duplicate prompts, keep first occurrence
df = df.drop_duplicates(subset=["prompt"], keep="first").reset_index(drop=True)

# Stats
total_after = len(df)
benign_count = int((df["attack_type"] == "").sum())
malicious_count = int((df["attack_type"] != "").sum())

# Save cleaned dataset
df.to_csv(out_path, index=False)

print(f"Total rows before cleaning: {total_before}")
print(f"Total rows after cleaning: {total_after}")
print(f"Number of benign rows: {benign_count}")
print(f"Number of malicious rows: {malicious_count}")
df.head()
```
